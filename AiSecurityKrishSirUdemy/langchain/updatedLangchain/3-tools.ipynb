### Tools

```python
import os 
from dotenv import load_dotenv
from langchain.chat_models import init_chat_model

load_dotenv()

os.environ["OPENAI_API_KEY"] = os.getenv("OPENAI_API_KEY")
os.environ["GOOGLE_API_KEY"] = os.getenv("GOOGLE_API_KEY")
os.environ["GROQ_API_KEY"] = os.getenv("GROQ_API_KEY")

## qwen using groq model 
model_groq = init_chat_model("groq:qwen/qwen3.6-27b", temperature=0.7)

```

```python
from langchain.tools import tool

@tool
def get_current_weather(location: str) -> str:
    """Get the current weather in a given location"""
    return f"The current weather in {location} is sunny with a temperature of 25 degrees Celsius."

model_with_tools = model_groq.bind_tools([get_current_weather])

res = model_with_tools.invoke("What is the current weather in New York?")
print(res)

for tool_call in res.tool_calls:
    print(f"Tool called: {tool_call['name']}")
    print(f"Tool output: {tool_call['args']}")
```
