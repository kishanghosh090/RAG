Models Integration With OpenAI, Google Gemini and GROQ

```python
import os 
from dotenv import load_dotenv
from langchain.chat_models import init_chat_model
load_dotenv()

os.environ["OPENAI_API_KEY"] = os.getenv("OPENAI_API_KEY")
os.environ["GOOGLE_API_KEY"] = os.getenv("GOOGLE_API_KEY")
os.environ["GROQ_API_KEY"] = os.getenv("GROQ_API_KEY")
```

```python

model_gpt4 = init_chat_model("gpt-4", temperature=0.7)
model_gemini = init_chat_model("google_genai:gemini-2.5-flash", temperature=0.7)
```

```python
res=model_gemini.invoke("how are you?")
res.content
```

```python
## qwen using groq model 
model_groq = init_chat_model("groq:qwen/qwen3.6-27b", temperature=0.7)
res=model_groq.invoke("how are you?")
res.content
```
