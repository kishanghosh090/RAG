### Messages



```python
import os
from langchain.chat_models import init_chat_model

os.environ["GROQ_API_KEY"] = os.getenv("GROQ_API_KEY")

model_groq = init_chat_model("groq:qwen/qwen3.6-27b", temperature=0.7)
```

```python
# text prompt
model_groq.invoke("please tell me a joke about AI and cybersecurity")
```

```python
model_groq
```

```python
from langchain.messages import SystemMessage, HumanMessage, AIMessage

messages = [
    SystemMessage(content="You are a helpful assistant that knows about programming."),
    HumanMessage(content="I love programming."),
    AIMessage(content="J'adore programmer."),
]
response = model_groq.invoke(messages)
response
```

```python
system_arg = SystemMessage("You are a helpful coding assistant that knows about Python programming.")

messages = [
    system_arg,
    HumanMessage(content="I love programming."),
]
response = model_groq.invoke(messages)
print(response.content)
```

```python
# detailed example of using messages with a system message and a human message

from langchain.messages import SystemMessage, HumanMessage

system_msg = SystemMessage("""
You are a helpful assistant that knows about programming. You will be given a question about programming, and you will answer it in a clear and concise manner. If you don't know the answer, you will say "I don't know."
""")
messages = [
    system_msg,
    HumanMessage(content="What is the difference between a list and a tuple in Python?"),
]


response = model_groq.invoke(messages)
print(response.content)
```

```python
human_msg = HumanMessage(
    content="What is the difference between a list and a tuple in Python?", 
    name="user1",
    id="12345"
)

res = model_groq.invoke([ human_msg])
print(res.content)
```
